# Procesamiento de Archivos de Posiciones

Este notebook procesa cuatro archivos de entrada para generar archivos de salida específicos.

## Archivos de entrada:
- HEU-NaPosiciones.dat
- HEU-SiOPosiciones.dat
- Texto.txt
- MejoresConfNakata-5.4C1.dat

```python
# Paso 1: Capturar las posiciones de los símbolos "Al" en cada línea del archivo MejoresConfNakata-5.4C1.dat

def obtener_posiciones_Al(archivo_entrada):
    """
    Lee el archivo de entrada y captura las posiciones (índices) donde se encuentran
    los símbolos "Al" para cada línea.
    
    Args:
        archivo_entrada: Ruta al archivo de entrada
    
    Returns:
        Lista de listas, donde cada sublista contiene las posiciones de "Al" para esa línea
    """
    posiciones_por_linea = []
    
    with open(archivo_entrada, 'r') as f:
        for num_linea, linea in enumerate(f, 1):
            # Dividir la línea en símbolos individuales
            simbolos = linea.strip().split()
            
            # Encontrar las posiciones donde está "Al"
            posiciones_Al = [indice for indice, simbolo in enumerate(simbolos) if simbolo == 'Al']
            
            posiciones_por_linea.append(posiciones_Al)
            print(f"Línea {num_linea}: Posiciones de Al = {posiciones_Al}")
    
    return posiciones_por_linea

# Ejecutar el paso 1
archivo_nakata = 'MejoresConfNakata-5.4C1.dat'
posiciones_Al = obtener_posiciones_Al(archivo_nakata)

print(f"\nTotal de líneas procesadas: {len(posiciones_Al)}")
print(f"Primeras 5 líneas con sus posiciones de Al:")
for i in range(min(5, len(posiciones_Al))):
    print(f"  Línea {i+1}: {posiciones_Al[i]}")
```

```python
# Paso 2: Procesar HEU-SiOPosiciones.dat y HEU-NaPosiciones.dat


def procesar_HEU_SiO_Posiciones(archivo_entrada):
    """
    Procesa el archivo HEU-SiOPosiciones.dat considerando solo los primeros 108 renglones.
    - Los primeros 72 renglones empiezan con 'O' (oxígeno)
    - Los siguientes 36 renglones empiezan con 'Si' (silicio)
    - Cada renglón tiene: símbolo (O o Si) + 3 valores numéricos
    
    Args:
        archivo_entrada: Ruta al archivo HEU-SiOPosiciones.dat
    
    Returns:
        Tupla con (lista_O, lista_Si) donde cada una contiene [símbolo, val1, val2, val3]
    """
    datos_O = []
    datos_Si = []
    
    with open(archivo_entrada, 'r') as f:
        for num_linea, linea in enumerate(f, 1):
            if num_linea > 108:
                break
            
            partes = linea.strip().split()
            if len(partes) >= 4:
                simbolo = partes[0]
                val1 = float(partes[1])
                val2 = float(partes[2])
                val3 = float(partes[3])
                
                if num_linea <= 72:
                    # Primeros 72 renglones: carácter 'O'
                    datos_O.append([simbolo, val1, val2, val3])
                else:
                    # Siguientes 36 renglones: carácter 'Si'
                    datos_Si.append([simbolo, val1, val2, val3])
    
    return datos_O, datos_Si



def procesar_HEU_Na_Posiciones(archivo_entrada):
    """
    Procesa el archivo HEU-NaPosiciones.dat (36 renglones).
    - Ignora las primeras 7 columnas y todo lo más allá de la columna 40
    - Captura los caracteres 'Na' y los 3 valores numéricos subsecuentes
    
    Formato típico: Na1    Na    0.29474   0.07838   0.02060   0.00000  Uiso   1.00
    Se captura: 'Na' y los 3 valores numéricos siguientes
    
    Args:
        archivo_entrada: Ruta al archivo HEU-NaPosiciones.dat
    
    Returns:
        Lista de listas, donde cada una contiene [símbolo 'Na', val1, val2, val3]
    """
    datos_Na = []
    
    with open(archivo_entrada, 'r') as f:
        for num_linea, linea in enumerate(f, 1):
            # Extraer solo hasta la columna 40 (índice 0-39)
            linea_truncada = linea[:40]
            
            # Dividir en partes
            partes = linea_truncada.strip().split()
            
            # Buscar el símbolo 'Na' y los 3 valores siguientes
            for i, parte in enumerate(partes):
                if parte == 'Na' and i + 3 < len(partes):
                    try:
                        val1 = float(partes[i+1])
                        val2 = float(partes[i+2])
                        val3 = float(partes[i+3])
                        datos_Na.append(['Na', val1, val2, val3])
                        break
                    except ValueError:
                        continue
    
    return datos_Na



# Ejecutar el paso 2
archivo_SiO = 'HEU-SiOPosiciones.dat'
archivo_Na = 'HEU-NaPosiciones.dat'



# Procesar HEU-SiOPosiciones.dat
datos_O, datos_Si = procesar_HEU_SiO_Posiciones(archivo_SiO)
print(f"HEU-SiOPosiciones.dat:")
print(f"  - Número de átomos O procesados: {len(datos_O)}")
print(f"  - Número de átomos Si procesados: {len(datos_Si)}")
print(f"  - Primeros 3 átomos O: {datos_O[:3]}")
print(f"  - Primeros 3 átomos Si: {datos_Si[:3]}")



# Procesar HEU-NaPosiciones.dat
datos_Na = procesar_HEU_Na_Posiciones(archivo_Na)
print(f"\nHEU-NaPosiciones.dat:")
print(f"  - Número de átomos Na procesados: {len(datos_Na)}")
print(f"  - Primeros 3 átomos Na: {datos_Na[:3]}")
```
